```python
import pandas as pd
import numpy as np
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
def binary_cross_entropy(y_true, y_pred):
    epsilon = 1e-8  
    return -np.mean(y_true * np.log(y_pred + epsilon) + (1 - y_true) * np.log(1 - y_pred + epsilon))


def initialize_weights(n_features):
    weights = np.random.randn(n_features) * 0.01
    bias = 0.0
    return weights, bias

def propagate(X, y, weights, bias):
    m = X.shape[0] 
    
 
    z = np.dot(X, weights) + bias
    y_pred = sigmoid(z)
    loss = binary_cross_entropy(y, y_pred)
    

    dw = (1 / m) * np.dot(X.T, (y_pred - y))
    db = (1 / m) * np.sum(y_pred - y)
    
    return dw, db, loss


def optimize(X, y, weights, bias, learning_rate, epochs):
    for epoch in range(epochs):
        dw, db, loss = propagate(X, y, weights, bias)
        

        weights -= learning_rate * dw
        bias -= learning_rate * db
        

        if epoch % 100 == 0:
            print(f"Epoch {epoch}, Loss: {loss:.4f}")
    
    return weights, bias


def predict(X, weights, bias):
    z = np.dot(X, weights) + bias
    y_pred = sigmoid(z)
    return (y_pred > 0.5).astype(int)

file_path = 'dynamic_pricing_dataset.csv'
data = pd.read_csv(file_path)


data['Label'] = np.where(
    (data['Demand_Index'] > 5) & (data['Price'] < 1500),
    "success",
    "fail"
)


features = data[['Advertising_Spend', 'Competitor_Price', 'Demand_Index', 'Price']].values
labels = (data['Label'] == "success").astype(int)  # Convert 'success' to 1 and 'fail' to 0


n_features = features.shape[1]
weights, bias = initialize_weights(n_features)
learning_rate = 0.01
epochs = 1000


trained_weights, trained_bias = optimize(features, labels, weights, bias, learning_rate, epochs)


predictions = predict(features, trained_weights, trained_bias)
accuracy = np.mean(predictions == labels) * 100

print(f"Training Accuracy: {accuracy:.2f}%")
```
